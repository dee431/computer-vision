```python
from IPython.display import Javascript
from google.colab.output import eval_js
from base64 import b64decode

def take_photo(filename='photo.jpg', quality=0.8):

    js = Javascript('''
    async function takePhoto(quality) {

      const div = document.createElement('div');
      const capture = document.createElement('button');
      capture.textContent = 'Capture';

      div.appendChild(capture);

      const video = document.createElement('video');

      video.style.display = 'block';

      const stream =
          await navigator.mediaDevices.getUserMedia({video: true});

      document.body.appendChild(div);
      div.appendChild(video);

      video.srcObject = stream;

      await video.play();

      google.colab.output.setIframeHeight(document.documentElement.scrollHeight,true);

      await new Promise((resolve) => capture.onclick = resolve);

      const canvas = document.createElement('canvas');

      canvas.width = video.videoWidth;

      canvas.height = video.videoHeight;

      canvas.getContext('2d').drawImage(video,0,0);

      stream.getVideoTracks()[0].stop();

      div.remove();

      return canvas.toDataURL('image/jpeg',quality);

    }
    ''')

    display(js)

    data = eval_js('takePhoto({})'.format(quality))

    binary = b64decode(data.split(',')[1])

    with open(filename,'wb') as f:
        f.write(binary)

    return filename
```

**Cell 2: Open camera**

```python
from IPython.display import Javascript, display
from google.colab.output import eval_js
from base64 import b64decode

def open_camera(filename="photo.jpg", quality=0.9):
    js = Javascript("""
    async function openCamera(quality){
        const div = document.createElement('div');
        const openBtn = document.createElement('button');
        openBtn.innerHTML = "📷 Open Camera";
        openBtn.style.fontSize = "20px";
        openBtn.style.padding = "12px 20px";
        openBtn.style.margin = "10px";
        const captureBtn = document.createElement('button');
        captureBtn.innerHTML = "📸 Capture Image";
        captureBtn.style.fontSize = "20px";
        captureBtn.style.padding = "12px 20px";
        captureBtn.style.margin = "10px";
        captureBtn.style.display = "none";
        div.appendChild(openBtn);
        div.appendChild(captureBtn);
        document.body.appendChild(div);
        const video = document.createElement('video');
        video.width = 640;
        video.style.display = "none";
        div.appendChild(video);
        let stream;
        await new Promise(resolve => {
            openBtn.onclick = async () => {
                stream = await navigator.mediaDevices.getUserMedia({video:true});
                video.srcObject = stream;
                video.style.display = "block";
                captureBtn.style.display = "inline";
                await video.play();
                google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);
                resolve();
            }
        });
        await new Promise(resolve=>{captureBtn.onclick=resolve;});
        const canvas=document.createElement('canvas');
        canvas.width=video.videoWidth;
        canvas.height=video.videoHeight;
        canvas.getContext('2d').drawImage(video,0,0);
        stream.getTracks().forEach(track=>track.stop());
        const data=canvas.toDataURL('image/jpeg',quality);
        div.remove();
        return data;
    }
    """)
    display(js)
    data = eval_js("openCamera({})".format(quality))
    binary = b64decode(data.split(",")[1])
    with open(filename,"wb") as f:
        f.write(binary)
    return filename

image_path = open_camera()
print("✅ Image saved successfully!")
print("Image Path:", image_path)
```

**Detect Objects**

```python
from ultralytics import YOLO

# Ensure model is loaded to avoid NameError
if 'model' not in globals():
    model = YOLO('yolov8n.pt')

results = model.predict(
    image_path,
    conf=0.50,
    verbose=False
)
```

**Cell 7 – Debug the Detection**

```python
boxes = results[0].boxes

print("Number of detections:", len(boxes))

for box in boxes:

    cls = int(box.cls)
    conf = float(box.conf)

    print("Detected:", model.names[cls])
    print("Confidence:", conf)
```

**Get Object Names**

```python
names = model.names

boxes = results[0].boxes

detected_objects = []

for box in boxes:

    cls = int(box.cls)

    detected_objects.append(names[cls])

print(detected_objects)
```

**Convert Text to Speech**

```python
from gtts import gTTS
from IPython.display import Audio

sentence = "I can see "

if detected_objects:
    sentence += ", ".join(detected_objects)
else:
    sentence += "nothing clear right now."

tts = gTTS(sentence)
tts.save("voice.mp3")

Audio("voice.mp3", autoplay=True)
```

**Real-Time Loop**

```python
count = 0
max_captures = 3 # Limiting to 3 captures for demonstration

while count < max_captures:

    print(f"\n--- Capture {count + 1} of {max_captures} ---")
    image = take_photo()

    results = model(image)

    boxes = results[0].boxes

    names = model.names

    detected=[]

    for box in boxes:

        cls=int(box.cls)

        detected.append(names[cls])

    if len(detected)>0:

        sentence="I can see "+", ".join(detected)

        print(sentence)

        tts=gTTS(sentence)

        tts.save("voice.mp3")

        display(Audio("voice.mp3", autoplay=True))

    else:
        print("No objects detected in this capture.")

    count += 1
    if count < max_captures:
        again=input(f"Continue for {max_captures - count} more captures? (y/n): ")
        if again.lower()!="y":
            break
    else:
        print("Maximum captures reached. Loop terminated.")

print("\nReal-time loop finished.")
```
